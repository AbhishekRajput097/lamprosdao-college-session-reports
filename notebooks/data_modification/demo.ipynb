```python
import pandas as pd
import random
import re

def generate_email(full_name):
    """
    Generate a realistic email address based on the full name with varied patterns.
    """
    name_parts = re.split(r'\s+', full_name.strip().lower())  # Split full name into parts

    if len(name_parts) == 1:
        first_name = name_parts[0]
        last_name = ""
    else:
        first_name, last_name = name_parts[0], name_parts[-1]

    random_number = random.randint(1, 9999)  # Generate a random number between 1 and 4 digits

    # Define possible email patterns
    patterns = [
        f"{first_name}@gmail.com",
        f"{last_name}@gmail.com",
        f"{first_name}{last_name}@gmail.com",
        f"{first_name}{random_number}@gmail.com",
        f"{last_name}{random_number}@gmail.com",
        f"{first_name}{last_name}{random_number}@gmail.com"
    ]

    # Randomly choose a pattern
    email = random.choice(patterns)
    return email

def update_email_addresses(input_file, output_file):
    """
    Read the CSV file, generate new email addresses, and save the updated file.
    """
    # Load the CSV file
    df = pd.read_csv(input_file)

    # Generate new email addresses
    df['Email address'] = df['Full Name'].apply(generate_email)

    # Save the updated data to a new CSV file
    df.to_csv(output_file, index=False)
    print(f"Updated CSV file saved to {output_file}")

# Input and output file names
input_csv = "../../data/data_modification/session_dummy_database.csv"
output_csv = "../../data/data_modification/updated_session_dummy_database.csv"

# Update the email addresses
update_email_addresses(input_csv, output_csv)
```

```python
import pandas as pd
import random
from datetime import datetime, timedelta

# Load both CSV files
file_one = "../../data/data_modification/session_dummy_database_one.csv"
file_updated = "../../data/data_modification/updated_session_dummy_database.csv"
data_one = pd.read_csv(file_one)
data_updated = pd.read_csv(file_updated)

# Function to generate a random timestamp within a given range
def random_timestamp(start_time, end_time):
    start = datetime.strptime(start_time, "%d-%m-%Y %H:%M")
    end = datetime.strptime(end_time, "%d-%m-%Y %H:%M")
    random_time = start + timedelta(seconds=random.randint(0, int((end - start).total_seconds())))
    return random_time.strftime("%d-%m-%Y %H:%M")

# List of unique colleges in the first file
colleges = data_one["College Name"].unique()

# Create a new DataFrame to store merged data
merged_data = pd.DataFrame()

# Process each college
for college in colleges:
    # Filter data for the current college
    data_one_college = data_one[data_one["College Name"] == college]
    data_updated_college = data_updated[data_updated["College Name"] == college]

    # Determine timestamp range for the current college in the first file
    timestamps = pd.to_datetime(data_one_college["Timestamp"], format="%d-%m-%Y %H:%M")
    if not timestamps.empty:
        start_time = timestamps.min().strftime("%d-%m-%Y %H:%M")
        end_time = timestamps.max().strftime("%d-%m-%Y %H:%M")

        # Update timestamps in the updated data for this college
        data_updated_college["Timestamp"] = data_updated_college["Timestamp"].apply(
            lambda _: random_timestamp(start_time, end_time)
        )

    # Append original and updated data for the current college
    merged_data = pd.concat([merged_data, data_one_college, data_updated_college], ignore_index=True)

# Save the merged data to a new CSV file
output_file = "../../data/data_modification/merged_session_dummy_database.csv"
merged_data.to_csv(output_file, index=False)
print(f"Merged data saved to {output_file}")
```
