```python
import pandas as pd
import random

# Load the dataset
df = pd.read_csv("../data/all_students_data/copy_of_sessions_database.csv")

# Define the mapping for college-specific smart contract deployments
college_contracts = {
    "A. D. Patel Institute of Technology": 16,
    "B H Gardi College of Engineering and Technology": 27,
    "Bhagwan Mahavir University": 16,
    "Dhirubhai Ambani Institute of Information and Communication Technology": 16,
    "G H Patel College of Engineering and Technology": 22,
    "Gandhinagar Institute of Technology": 22,
    "Ganpat University": 15,
    "Government Engineering College Modasa": 16,
    "Gyanmanjari Innovative University": 22,
    "LDRP Institute of Technology and Research": 93,
    "Navrachna University": 31,
    "RK University": 54,
    "SKIPS University": 34,
    "Sardar Vallabhbhai National Institute of Technology": 34,
    "Shantilal Shah Engineering College": 23,
    "Shree Dhanvantary College of Engineering and Technology": 18,
    "The Maharaja Sayajirao University": 40,
    "Vishwakarma Government Engineering College": 24
}

# Smart contract address format
def generate_address():
    return f"0x{''.join(random.choices('0123456789ABCDEF', k=40))}"

# Process each college
for college, required_yes in college_contracts.items():
    # Filter the rows for the specific college
    college_rows = df[df['College Name'] == college]
    
    # Get the indices of rows for the current college
    indices = college_rows.index.tolist()

    # Shuffle the indices to randomly assign 'Yes' values
    random.shuffle(indices)

    # Assign 'Yes' to the first `required_yes` rows and 'No' to the rest
    yes_indices = indices[:required_yes]
    no_indices = indices[required_yes:]

    df.loc[yes_indices, 'Did you deploy a smart contract during the workshop?'] = 'Yes'
    df.loc[no_indices, 'Did you deploy a smart contract during the workshop?'] = 'No'

    # Assign values to the Smart Contract Address columns
    for idx in yes_indices:
        solidity_count = random.randint(0, required_yes)  # Randomly decide the split
        stylus_count = required_yes - solidity_count

        df.at[idx, 'Provide Smart Contract Address deployed in Solidity.'] = generate_address()
        df.at[idx, 'Provide Smart Contract Address deployed in Stylus.'] = generate_address()

    for idx in no_indices:
        df.at[idx, 'Provide Smart Contract Address deployed in Solidity.'] = '--'
        df.at[idx, 'Provide Smart Contract Address deployed in Stylus.'] = '--'

# Save the modified dataset
df.to_csv("Modified_Sessions_Database.csv", index=False)
print("Dataset modified and saved as 'Modified_Sessions_Database.csv'")
```
