```python
import pandas as pd

# Load the CSV file
file_path = "../data/all_students_data/copy_of_sessions_database.csv"  # Replace with the actual path to your CSV file
data = pd.read_csv(file_path)

# Replace "--" with "No" for consistency (optional, based on your data)
data.replace("--", "No", inplace=True)

# Group by 'College Name' and calculate the required metrics
result = data.groupby('College Name').agg(
    Male_Count=('Gender', lambda x: (x == 'Male').sum()),
    Female_Count=('Gender', lambda x: (x == 'Female').sum()),
    Smart_Contract_Deployed=('Did you deploy a smart contract during the workshop?', lambda x: (x == 'Yes').sum()),
    Arbitrum_Accounts_Created=('Were you able to create your account on Arbitrum Forum?', lambda x: (x == 'Yes').sum())
).reset_index()

# Save the result to a new CSV file
output_file = "../data/college_analysis.csv"
result.to_csv(output_file, index=False)

print("Analysis results have been saved to", output_file)
print(result)
```
