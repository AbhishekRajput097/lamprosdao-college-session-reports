```python
import pandas as pd
import random

# Load the CSV file into a DataFrame
file_path = "../../data/all_students_data/copy_of_sessions_database.csv"
df = pd.read_csv(file_path)

# Define the possible values and their weights
courses = [
    "B.Sc. (Bachelor of Science)",
    "M.Sc. (Master of Science)",
    "B.Tech. (Bachelor of Technology)",
    "M.Tech. (Master of Technology)",
    "B.E. (Bachelor of Engineering)",
    "M.E. (Master of Engineering)",
    "Ph.D. / Faculty",
    "BCA (Bachelor of Computer Applications)",
    "MCA (Master of Computer Applications)",
    "Diploma in Engineering",
    "Postgraduate Diploma in Engineering"
]

# Define the weights (higher for bachelor's degrees, medium for master's, and lower for Ph.D.)
weights = [
    25,  # B.Sc.
    10,  # M.Sc.
    25,  # B.Tech.
    10,  # M.Tech.
    20,  # B.E.
    10,  # M.E.
    5,   # Ph.D.
    15,  # BCA
    10,  # MCA
    10,  # Diploma
    10   # Postgraduate Diploma
]

# Add random values to the 'Course/Department' column based on weights
df['Course/Department'] = [random.choices(courses, weights=weights, k=1)[0] for _ in range(len(df))]

# Save the updated DataFrame to a new CSV file
output_file_path = "../../data/all_students_data/updated_sessions_database.csv"
df.to_csv(output_file_path, index=False)

print(f"Updated file saved to {output_file_path}")
```
