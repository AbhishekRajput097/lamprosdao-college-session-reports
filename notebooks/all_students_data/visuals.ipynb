```python
import pandas as pd
import plotly.express as px

# Load the data
file_path = "../../data/all_students_data/copy_of_sessions_database.csv"
df = pd.read_csv(file_path)

# Ensure column name matches (case-sensitive)
gender_distribution = df['Gender'].value_counts().reset_index()
gender_distribution.columns = ['Gender', 'Count']

# Total number of attendees
total_attendees = 1285

# Create the pie chart with enhanced interactivity
fig = px.pie(
    gender_distribution,
    names='Gender',
    values='Count',
    title='<b>Gender-wise Distribution</b>',
    color_discrete_sequence=['#123869', '#12AAFF', '#BDD5F4'],
    hole=0.5  # Increased hole size
)

# Add customization
fig.update_traces(
    textinfo='percent+label',  # Show percentage and labels
    hovertemplate='<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}<extra></extra>',
    marker=dict(line=dict(color='#FFFFFF', width=2))  # Add a border for slices
)

# Add annotation for the center text
fig.add_annotation(
    text=f"<b>Total Attendees<br>{total_attendees}</b>",  # Add bold text for clarity
    x=0.5,  # Center horizontally
    y=0.5,  # Center vertically
    font=dict(size=18, color='#123869'),  # Adjusted font size
    showarrow=False,  # No arrow
    align='center'  # Center align text
)

# Update layout for better appearance and increased size
fig.update_layout(
    title=dict(
        x=0.5,
        xanchor='center',
        font=dict(size=20, color='#123869')
    ),
    font=dict(family="Arial", size=14, color='#123869'),
    margin=dict(t=80, b=80, l=80, r=80),  # Adjust margins for better spacing
    width=800,  # Increase width
    height=800  # Increase height
)

# Show the plot
fig.show()
# fig.write_html("Gender Wise Distribution.html")
```

```python
import pandas as pd
import plotly.express as px

# Load the data
file_path = "../../data/all_students_data/updated_sessions_database.csv"
df = pd.read_csv(file_path)

# Ensure column name matches (case-sensitive)
course_distribution = df['Course/Department'].value_counts().reset_index()
course_distribution.columns = ['Course/Department', 'Count']

# Calculate the total count and percentage for each course
total_count = course_distribution['Count'].sum()
course_distribution['Percentage'] = (course_distribution['Count'] / total_count * 100).round(2)

# Create the horizontal bar chart
fig = px.bar(
    course_distribution,
    x='Count',
    y='Course/Department',
    orientation='h',  # Horizontal bars
    title='<b>Course-wise Distribution</b>',
    color='Count',  # Use color to represent count for a heatmap-like effect
    color_continuous_scale=['#BDD5F4', '#12AAFF', '#123869'],  # Custom color scheme
    text='Percentage'  # Display percentage on the bars
)

# Customize the layout for aesthetics
fig.update_layout(
    title=dict(
        x=0.5,
        xanchor='center',
        font=dict(size=20, color='#123869')
    ),
    xaxis_title='Number of Attendees',
    yaxis_title='Course/Department',
    font=dict(family="Arial", size=14, color='#123869'),
    margin=dict(t=50, b=50, l=120, r=50),  # Adjust margins for readability
)

# Customize hover information and bar text
fig.update_traces(
    hovertemplate='<b>Course:</b> %{y}<br><b>Count:</b> %{x}<br><b>Percentage:</b> %{text}%<extra></extra>',
    texttemplate='%{text}%',  # Show percentage on the bars
    textposition='outside'  # Position text outside the bars
)

# Show the plot
fig.show()
fig.write_html("Course Wise Distribution.html")
```

```python
import pandas as pd
import plotly.graph_objects as go

# Load the data
file_path = "../../data/all_students_data/copy_of_sessions_database.csv"
df = pd.read_csv(file_path)

# Ensure the column name matches (case-sensitive)
rating_column = 'Rate the overall quality of the workshop'

# Count the occurrences of each rating
rating_distribution = df[rating_column].value_counts().reset_index()
rating_distribution.columns = ['Rating', 'Count']
rating_distribution = rating_distribution.sort_values(by='Rating')  # Sort ratings in ascending order

# Calculate percentages
total_count = rating_distribution['Count'].sum()
rating_distribution['Percentage'] = (rating_distribution['Count'] / total_count * 100).round(2)

# Define custom text positions to avoid overlaps
text_positions = []
for index, row in rating_distribution.iterrows():
    if row['Count'] == 45:  # Custom adjustment for count 45
        text_positions.append('bottom center')  # Move text below
    elif row['Count'] <= 50:  # Small counts
        text_positions.append('top right')
    else:  # Default for larger counts
        text_positions.append('top center')

# Create the lollipop chart
fig = go.Figure()

# Add the stems (lines)
fig.add_trace(go.Scatter(
    x=rating_distribution['Rating'],
    y=rating_distribution['Count'],
    mode='lines',
    line=dict(color='#BDD5F4', width=2),
    showlegend=False
))

# Add the dots (lollipop heads)
fig.add_trace(go.Scatter(
    x=rating_distribution['Rating'],
    y=rating_distribution['Count'],
    mode='markers+text',
    marker=dict(color='#123869', size=10),
    text=[f"{count} ({perc}%)" for count, perc in zip(rating_distribution['Count'], rating_distribution['Percentage'])],
    textposition=text_positions,  # Use custom text positions
    textfont=dict(size=12, color='#123869'),
    showlegend=False
))

# Customize the layout for aesthetics
fig.update_layout(
    title='<b>Rating Distribution: Quality of the Session</b>',
    title_x=0.5,
    xaxis=dict(
        title='Rating (1-5)',
        tickmode='linear',
        tick0=1,
        dtick=1
    ),
    yaxis=dict(
        title='Number of Attendees'
    ),
    font=dict(family="Arial", size=14, color='#123869'),
    margin=dict(t=50, b=50, l=50, r=50),
    plot_bgcolor='rgba(0,0,0,0)',  # Transparent background
)

# Show the plot
fig.show()
fig.write_html("Rating Distribution_Quality of Session.html")
```

```python
import pandas as pd
import plotly.express as px

# Load the data
file_path = "../../data/all_students_data/copy_of_sessions_database.csv"
df = pd.read_csv(file_path)

# Ensure the column name matches (case-sensitive)
rating_column = "Rate the instructor’s knowledge and presentation skills."

# Count the occurrences of each rating
rating_distribution = df[rating_column].value_counts().reset_index()
rating_distribution.columns = ["Rating", "Count"]
rating_distribution = rating_distribution.sort_values(by="Rating")  # Sort ratings in ascending order

# Calculate percentages
rating_distribution['Percentage'] = (rating_distribution['Count'] / rating_distribution['Count'].sum() * 100).round(2)

# Create the pie chart
fig = px.pie(
    rating_distribution,
    names='Rating',
    values='Count',
    title="<b>Rating Distribution: Instructor's Knowledge</b>",
    color='Rating',
    color_discrete_sequence=["#BDD5F4", "#12AAFF", "#90CAF9", "#1E88E5", "#123869"]  # Custom colors
)

# Customize text and layout
fig.update_traces(
    text=[f"Score: {rating}<br>({perc}%)" for rating, perc in zip(rating_distribution['Rating'], rating_distribution['Percentage'])],  # Label with rating and percentage
    textinfo='text',  # Display label and percentage
    textposition='inside',  # Position text inside the pie slices
    insidetextorientation='horizontal',  # Ensure text is horizontal
    hovertemplate='<b>Rating:</b> %{label}<br><b>Count:</b> %{value}<br><b>Percentage:</b> %{percent}<extra></extra>',
    marker=dict(line=dict(color='#FFFFFF', width=1))  # Add white borders for clarity
)

# Customize layout with increased size
fig.update_layout(
    title=dict(
        x=0.5,
        xanchor="center",
        font=dict(size=20, color="#123869")
    ),
    font=dict(family="Arial", size=14, color="#123869"),
    margin=dict(t=50, b=50, l=50, r=50),
    showlegend=False,  # Hide legend since labels are inside
    width=900,  # Increased width
    height=700  # Increased height
)

# Show the plot
fig.show()
fig.write_html("Rating Distribution_Instructor's Knowledge.html")
```
